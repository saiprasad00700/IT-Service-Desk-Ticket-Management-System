# IT Service Desk and Ticket Management System

## Project Overview

This project is a Python-based IT Service Desk and Ticket Management System
integrated with a MySQL database.

The system allows users to create and manage IT support tickets, assign
tickets to support agents, update ticket priority and status, search tickets,
and generate support insights using SQL queries.

## Key Features

- Create IT support tickets
- Assign tickets to support agents
- Update ticket status and priority
- Search and filter tickets
- Track ticket resolution
- Analyze ticket status and priority
- Monitor agent workload
- Calculate average ticket resolution time

```python
def update_ticket():

    ticket_id = int(input("Enter Ticket ID: "))

    query = """
    SELECT ticket_id
    FROM tickets
    WHERE ticket_id = %s
    """

    cursor.execute(query, (ticket_id,))
    ticket = cursor.fetchone()

    if ticket is None:
        print("Ticket not found.")
        return

    status = input(
        "Enter new status (Open/In Progress/Resolved/Closed): "
    )

    priority = input(
        "Enter new priority (Low/Medium/High/Critical): "
    )

    valid_statuses = [
        "Open",
        "In Progress",
        "Resolved",
        "Closed"
    ]

    valid_priorities = [
        "Low",
        "Medium",
        "High",
        "Critical"
    ]

    if status not in valid_statuses:
        print("Invalid status.")
        return

    if priority not in valid_priorities:
        print("Invalid priority.")
        return

    if status == "Resolved":

        query = """
        UPDATE tickets
        SET status = %s,
            priority = %s,
            resolved_at = CURRENT_TIMESTAMP
        WHERE ticket_id = %s
        """

    else:

        query = """
        UPDATE tickets
        SET status = %s,
            priority = %s,
            resolved_at = NULL
        WHERE ticket_id = %s
        """

    cursor.execute(
        query,
        (status, priority, ticket_id)
    )

    connection.commit()

    print("Ticket updated successfully!")
```

```python
def create_ticket():
    user_id = int(input("Enter User ID: "))
    title = input("Enter Issue Title: ")
    description = input("Enter Description: ")
    category = input("Enter Category: ")
    priority = input("Enter Priority (Low/Medium/High/Critical): ")

    valid_priorities = ["Low", "Medium", "High", "Critical"]

    if priority not in valid_priorities:
        print("Invalid priority.")
        return

    query = """
    INSERT INTO tickets
    (user_id, title, description, category, priority)
    VALUES (%s, %s, %s, %s, %s)
    """

    values = (user_id, title, description, category, priority)

    cursor.execute(query, values)
    connection.commit()

    ticket_id = cursor.lastrowid

    print("\nTicket created successfully!")
    print("Ticket ID:", ticket_id)
```

```python
def assign_ticket():
    ticket_id = int(input("Enter Ticket ID: "))
    agent_id = int(input("Enter Agent ID: "))

    # Check ticket
    query = """
    SELECT ticket_id
    FROM tickets
    WHERE ticket_id = %s
    """

    cursor.execute(query, (ticket_id,))
    ticket = cursor.fetchone()

    if ticket is None:
        print("Ticket not found.")
        return

    # Check agent
    query = """
    SELECT agent_id
    FROM agents
    WHERE agent_id = %s
    """

    cursor.execute(query, (agent_id,))
    agent = cursor.fetchone()

    if agent is None:
        print("Agent not found.")
        return

    # Assign agent
    query = """
    UPDATE tickets
    SET agent_id = %s
    WHERE ticket_id = %s
    """

    cursor.execute(query, (agent_id, ticket_id))
    connection.commit()

    print("Ticket assigned successfully!")
```

```python
def average_resolution_time():
    query = """
    SELECT
        AVG(
            TIMESTAMPDIFF(
                MINUTE,
                created_at,
                resolved_at
            )
        ) / 60 AS average_hours
    FROM tickets
    WHERE resolved_at IS NOT NULL
    """

    cursor.execute(query)
    result = cursor.fetchone()

    average_hours = result[0]

    if average_hours is not None:
        print(f"Average Resolution Time: {average_hours:.2f} hours")
    else:
        print("No resolved tickets available.")
```

```python
def agent_workload():
    query = """
    SELECT
        a.name AS agent_name,
        COUNT(t.ticket_id) AS ticket_count
    FROM agents a
    LEFT JOIN tickets t
        ON a.agent_id = t.agent_id
    GROUP BY a.agent_id, a.name
    ORDER BY ticket_count DESC
    """

    cursor.execute(query)
    results = cursor.fetchall()

    print("\nAgent Workload")
    print("-" * 30)

    for agent, count in results:
        print(f"{agent:<20} {count}")
```
